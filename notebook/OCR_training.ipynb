```python
import torch
from transformers import LayoutLMv3Processor, LayoutLMv3ForTokenClassification

checkpoint_path = "/content/drive/MyDrive/DocVision/layoutlmv3-cord"

processor = LayoutLMv3Processor.from_pretrained(
    checkpoint_path,
    apply_ocr=False
)

model = LayoutLMv3ForTokenClassification.from_pretrained(
    checkpoint_path
)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
model.eval()

print("Model loaded successfully")
print("Device:", device)
print("Number of labels:", model.config.num_labels)
print("Label mapping:", model.config.id2label)
```

```python
import numpy as np

result = results[0]
words = list(result["rec_texts"])
pixel_boxes = np.asarray(result["rec_boxes"])

width, height = image.size
boxes = []

for box in pixel_boxes:
    # Convert to Python integers BEFORE multiplication
    x1, y1, x2, y2 = [int(v) for v in box]

    # Normalize coordinates to the 0–1000 range
    x1 = max(0, min(1000, round(x1 * 1000 / width)))
    y1 = max(0, min(1000, round(y1 * 1000 / height)))
    x2 = max(0, min(1000, round(x2 * 1000 / width)))
    y2 = max(0, min(1000, round(y2 * 1000 / height)))

    # Ensure valid coordinate ordering
    x1, x2 = sorted([x1, x2])
    y1, y2 = sorted([y1, y2])

    boxes.append([x1, y1, x2, y2])

assert len(words) == len(boxes)
assert all(
    0 <= x1 <= x2 <= 1000 and 0 <= y1 <= y2 <= 1000
    for x1, y1, x2, y2 in boxes
)

print("First five corrected boxes:")
for word, box in zip(words[:5], boxes[:5]):
    print(word, box)
```

```python
encoding = processor(
    images=image.convert("RGB"),
    text=words,
    boxes=boxes,
    truncation=True,
    return_tensors="pt"
)

encoding = encoding.to(device)

with torch.no_grad():
    outputs = model(**encoding)

predictions = outputs.logits.argmax(dim=-1)[0].cpu().tolist()

print("Inference completed")
print("Predicted token count:", len(predictions))
```

```python
id2label = model.config.id2label

for token_id, prediction in enumerate(predictions[:30]):
    print(
        f"Token {token_id}: "
        f"{id2label[prediction]}"
    )
```

```python
import torch

# Get predicted class IDs
predictions = outputs.logits.argmax(dim=-1)[0].cpu().tolist()

# Get word IDs for each token
word_ids = encoding.word_ids(batch_index=0)

# Keep the first subtoken prediction for each OCR word
word_predictions = {}
for token_idx, word_id in enumerate(word_ids):
    if word_id is not None and word_id not in word_predictions:
        word_predictions[word_id] = predictions[token_idx]

# Display each OCR word with its predicted label
for word_id, word in enumerate(words):
    if word_id in word_predictions:
        label_id = word_predictions[word_id]
        label = model.config.id2label[label_id]
        print(f"{word:<25} -> {label}")
    else:
        print(f"{word:<25} -> No prediction")
```

```python
import json
import re

# Collect predicted words by entity label
entities = {}

for word_id, word in enumerate(words):
    if word_id not in word_predictions:
        continue

    label = model.config.id2label[word_predictions[word_id]]

    if label == "O":
        continue

    # Remove BIO prefix, e.g. B-menu.price -> menu.price
    entity = re.sub(r"^[BI]-", "", label)
    entities.setdefault(entity, []).append(word)

print(json.dumps(entities, indent=2, ensure_ascii=False))
```

```python
def first_value(entities, *keys):
    for key in keys:
        values = entities.get(key)
        if values:
            return " ".join(values)
    return None

receipt = {
    "items": entities.get("menu.nm", []),
    "quantities": entities.get("menu.cnt", []),
    "unit_prices": entities.get("menu.unitprice", []),
    "prices": entities.get("menu.price", []),
    "subtotal": first_value(
        entities, "sub_total.subtotal_price"
    ),
    "tax": first_value(
        entities, "sub_total.tax_price"
    ),
    "total": first_value(
        entities, "total.total_price"
    ),
}

print(json.dumps(receipt, indent=2, ensure_ascii=False))
```

```python
import re
import json

def extract_bio_entities(words, word_predictions, id2label):
    """
    Convert word-level BIO predictions into entity spans.

    B- starts a new entity.
    I- continues a matching entity.
    O closes the current entity.
    """
    entities = []
    current_entity = None

    def save_entity():
        if current_entity is not None:
            entities.append({
                "label": current_entity["label"],
                "text": " ".join(current_entity["words"]),
                "start_word": current_entity["start_word"],
                "end_word": current_entity["end_word"]
            })

    for word_id, word in enumerate(words):
        if word_id not in word_predictions:
            save_entity()
            current_entity = None
            continue

        label = id2label[int(word_predictions[word_id])]

        if label == "O":
            save_entity()
            current_entity = None
            continue

        match = re.match(r"^([BI])-(.+)$", label)

        if not match:
            save_entity()
            current_entity = None
            continue

        prefix, entity_type = match.groups()

        if prefix == "B":
            save_entity()
            current_entity = {
                "label": entity_type,
                "words": [word],
                "start_word": word_id,
                "end_word": word_id
            }

        elif prefix == "I":
            if (
                current_entity is not None
                and current_entity["label"] == entity_type
            ):
                current_entity["words"].append(word)
                current_entity["end_word"] = word_id
            else:
                # Handle an I- label without a matching preceding entity.
                save_entity()
                current_entity = {
                    "label": entity_type,
                    "words": [word],
                    "start_word": word_id,
                    "end_word": word_id
                }

    save_entity()
    return entities


entities = extract_bio_entities(
    words,
    word_predictions,
    model.config.id2label
)

print(json.dumps(entities, indent=2, ensure_ascii=False))
```

```python
from collections import defaultdict

grouped_entities = defaultdict(list)

for entity in entities:
    grouped_entities[entity["label"]].append(entity["text"])

grouped_entities = dict(grouped_entities)

print(json.dumps(grouped_entities, indent=2, ensure_ascii=False))
```

```python
import json
import numpy as np

ground_truth = sample["ground_truth"]
if isinstance(ground_truth, str):
    ground_truth = json.loads(ground_truth)

# Extract word-level ground-truth boxes and categories
gt_words = []

for line in ground_truth["valid_line"]:
    category = line["category"]

    for item in line["words"]:
        quad = item["quad"]

        xs = [quad[f"x{i}"] for i in range(1, 5)]
        ys = [quad[f"y{i}"] for i in range(1, 5)]

        gt_words.append({
            "text": item["text"],
            "label": category,
            "box": [
                min(xs), min(ys),
                max(xs), max(ys)
            ]
        })


def box_iou(a, b):
    ix1, iy1 = max(a[0], b[0]), max(a[1], b[1])
    ix2, iy2 = min(a[2], b[2]), min(a[3], b[3])

    intersection = max(0, ix2 - ix1) * max(0, iy2 - iy1)

    area_a = max(0, a[2] - a[0]) * max(0, a[3] - a[1])
    area_b = max(0, b[2] - b[0]) * max(0, b[3] - b[1])

    union = area_a + area_b - intersection
    return intersection / union if union > 0 else 0.0


comparison = []

for i, word in enumerate(words):
    pred_id = word_predictions.get(i)
    if pred_id is None:
        continue

    pred_label = model.config.id2label[pred_id]
    if pred_label == "O":
        pred_category = "O"
    else:
        pred_category = pred_label.split("-", 1)[-1]

    ocr_box = [int(v) for v in pixel_boxes[i]]

    # Find the ground-truth word with maximum spatial overlap
    best = max(
        gt_words,
        key=lambda item: box_iou(ocr_box, item["box"]),
        default=None
    )

    overlap = box_iou(ocr_box, best["box"]) if best else 0.0

    comparison.append({
        "ocr_text": word,
        "predicted": pred_category,
        "ground_truth": best["label"] if best and overlap > 0 else "unmatched",
        "iou": round(overlap, 3)
    })

print(json.dumps(comparison, indent=2, ensure_ascii=False))
```
